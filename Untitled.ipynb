```python
def extract_features(file_path, sr=22050):
    try:
        audio, sr = librosa.load(file_path, sr=sr, duration=5)
        audio = nr.reduce_noise(y=audio, sr=sr)
        audio, _ = librosa.effects.trim(audio, top_db=20)

        # SAFETY: skip files that are too short
        if len(audio) < 2000:  
            return None

        features = []

        # MFCC (40 coeffs)
        mfccs = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=40)
        features.extend(np.mean(mfccs, axis=1))
        features.extend(np.std(mfccs, axis=1))
        features.extend(np.max(mfccs, axis=1))

        # Delta MFCC
        delta = librosa.feature.delta(mfccs)
        features.extend(np.mean(delta, axis=1))
        features.extend(np.std(delta, axis=1))

        # Spectral features
        features.extend([
            np.mean(librosa.feature.spectral_centroid(y=audio, sr=sr)),
            np.mean(librosa.feature.spectral_rolloff(y=audio, sr=sr)),
            np.mean(librosa.feature.zero_crossing_rate(audio)),
            np.mean(librosa.feature.rms(y=audio))
        ])
        # Cough energy band (500–2000 Hz)
        sos = signal.butter(4, [500, 2000], btype='band', fs=sr, output='sos')
        filtered = signal.sosfilt(sos, audio)
        features.append(np.sum(filtered**2) / (np.sum(audio**2) + 1e-8))

        return np.array(features)

    except Exception as e:
        # print("Error:", e)   # optional debug
        return None
```

```python
import os
import numpy as np
from tqdm import tqdm

X, y = [], []

# Paths
cough_path = "final_dataset/cough"
non_cough_path = "final_dataset/non_cough"

# List files
cough_files = [f for f in os.listdir(cough_path) if f.endswith('.wav')]
non_cough_files = [f for f in os.listdir(non_cough_path) if f.endswith('.wav')]

print("Extracting COUGH features...")
for f in tqdm(cough_files):
    feat = extract_features(os.path.join(cough_path, f))
    if feat is not None:
        X.append(feat)
        y.append(1)   # cough label

print("Extracting NON-COUGH features...")
for f in tqdm(non_cough_files):
    feat = extract_features(os.path.join(non_cough_path, f))
    if feat is not None:
        X.append(feat)
        y.append(0)   # non-cough label

# Convert to arrays
X = np.array(X)
y = np.array(y)

# Save
os.makedirs("processed", exist_ok=True)
np.save("processed/features.npy", X)
np.save("processed/labels.npy", y)

print("\n=========================================")
print(f" Features extracted: {len(X)} samples")
print(f" Feature dimension: {X.shape[1]}")
print(" Saved to processed/features.npy and labels.npy")
print("=========================================")
print("Cell 13 Complete")
```

```python
X = np.load("processed/features.npy")
y = np.load("processed/labels.npy")

print(X.shape, y.shape)
```

```python
# Encode labels
le = LabelEncoder()
y_enc = le.fit_transform(y)

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y_enc, test_size=0.2, random_state=42, stratify=y_enc)

# Scale features
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)
```

```python


gb = GradientBoostingClassifier(
    n_estimators=200,
    learning_rate=0.1,
    subsample=0.8,
    max_depth=3   # valid parameter for GB trees
)

gb.fit(X_train_s, y_train)

# Predictions
y_pred = gb.predict(X_test_s)
acc = accuracy_score(y_test, y_pred)

print("\nTraining complete!")
print(f"Accuracy: {acc:.4f} ({acc*100:.2f}%)")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=["non_cough", "cough"]))

```
